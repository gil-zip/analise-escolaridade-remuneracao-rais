## Classificação dos setores econômicos

A atividade econômica dos vínculos é identificada por meio da
Classificação Nacional de Atividades Econômicas (CNAE 2.0).

A partir da divisão CNAE, os vínculos são agregados nos cinco grandes
setores utilizados na divulgação da RAIS:

- Agropecuária: seção A (divisões 01 a 03);
- Indústria: seções B a E (divisões 05 a 39);
- Construção: seção F (divisões 41 a 43);
- Comércio: seção G (divisões 45 a 47);
- Serviços: seções H a U (divisões 49 a 99).

A divisão CNAE corresponde aos dois primeiros dígitos do código da
classe CNAE 2.0.

Fonte: IBGE/CONCLA — Classificação Nacional de Atividades Econômicas
(CNAE 2.0).

```python
def obter_divisao_cnae(cnae):
    if pd.isna(cnae) or cnae == -1:
        return pd.NA

    try:
        codigo = str(int(cnae)).zfill(5)
        return int(codigo[:2])
    except (ValueError, TypeError):
        return pd.NA
```

```python
def classificar_setor(divisao):
    if pd.isna(divisao):
        return pd.NA

    if 1 <= divisao <= 3:
        return "Agropecuária"

    elif 5 <= divisao <= 39:
        return "Indústria"

    elif 41 <= divisao <= 43:
        return "Construção"

    elif 45 <= divisao <= 47:
        return "Comércio"

    elif 49 <= divisao <= 99:
        return "Serviços"

    return pd.NA
```

```python
# Testando as funções com alguns códigos CNAE
for codigo in [33147, 64638, 47814, 38114, 42227]:
    divisao = obter_divisao_cnae(codigo)

    print(
        codigo,
        divisao,
        classificar_setor(divisao)
    )
```

```python
import pyarrow as pa
import pyarrow.parquet as pq
```

### Processamento em chunks

A base completa é processada em blocos de 500.000 registros para evitar
o carregamento simultâneo de todo o arquivo na memória.

Em cada bloco são mantidos apenas vínculos ativos em 31/12, com
remuneração média nominal positiva, escolaridade conhecida e atividade
econômica classificável.

Os registros resultantes são gravados incrementalmente em formato
Parquet.

```python
arquivo_saida = DATA_PROCESSED / "rais_sp_2025.parquet"

chunksize = 500_000

contadores = {
    "originais": 0,
    "ativos": 0,
    "remuneracao_positiva": 0,
    "escolaridade_valida": 0,
    "setor_valido": 0,
    "base_final": 0
}

writer = None

try:
    for numero_chunk, chunk in enumerate(
        pd.read_csv(
            arquivo_rais,
            sep=",",
            encoding="latin1",
            usecols=colunas_processamento,
            chunksize=chunksize
        ),
        start=1
    ):
        # Total recebido neste bloco
        contadores["originais"] += len(chunk)

        # 1. Vínculos ativos em 31/12
        chunk = chunk[
            chunk["Ind Vínculo Ativo 31/12 - Código"] == 1
        ].copy()

        contadores["ativos"] += len(chunk)

        # 2. Escolaridade
        chunk["Escolaridade"] = (
            chunk["Escolaridade Após 2005 - Código"]
            .map(mapa_escolaridade)
        )

        # 3. Divisão CNAE e setor
        chunk["CNAE_Divisao"] = (
            chunk["CNAE 2.0 Classe - Código"]
            .apply(obter_divisao_cnae)
        )

        chunk["Setor"] = (
            chunk["CNAE_Divisao"]
            .apply(classificar_setor)
        )

        # Contagens antes dos filtros seguintes
        contadores["remuneracao_positiva"] += (
            chunk["Vl Rem Média Nom"] > 0
        ).sum()

        contadores["escolaridade_valida"] += (
            chunk["Escolaridade"].notna()
        ).sum()

        contadores["setor_valido"] += (
            chunk["Setor"].notna()
        ).sum()

        # 4. População analítica
        chunk = chunk[
            (chunk["Vl Rem Média Nom"] > 0) &
            (chunk["Escolaridade"].notna()) &
            (chunk["Setor"].notna())
        ].copy()

        contadores["base_final"] += len(chunk)

        # 5. Gravação incremental em Parquet
        tabela = pa.Table.from_pandas(
            chunk,
            preserve_index=False
        )

        if writer is None:
            writer = pq.ParquetWriter(
                arquivo_saida,
                tabela.schema,
                compression="snappy"
            )

        writer.write_table(tabela)

        print(
            f"Chunk {numero_chunk:02d} | "
            f"Lidos: {contadores['originais']:,} | "
            f"Base final: {contadores['base_final']:,}"
        )

finally:
    if writer is not None:
        writer.close()
```

## Resultado do processamento

A base original da RAIS 2025 de São Paulo continha 25.497.764 registros.

Após a seleção dos vínculos ativos em 31/12, foram identificados
16.365.074 vínculos. Destes, 14.948.065 apresentaram remuneração média
nominal positiva e compõem a base analítica final.

Todos os vínculos ativos apresentaram códigos de escolaridade mapeáveis
e atividade econômica classificável nos cinco grandes setores
considerados no estudo.

A base processada foi armazenada em formato Parquet, reduzindo o volume
necessário para as etapas posteriores e permitindo maior eficiência nas
análises exploratórias e estatísticas.

A unidade de análise permanece sendo o vínculo formal de trabalho, e
não o trabalhador individual.

Como validação externa, o número de vínculos ativos identificado nos
microdados (16.365.074) foi comparado ao estoque de empregos formais
divulgado oficialmente para São Paulo na RAIS 2025 (16.158.390).

Foi observada uma diferença de aproximadamente 1,28%. Como os microdados
e a divulgação oficial podem envolver critérios e tratamentos
metodológicos distintos, nenhum ajuste artificial foi realizado para
forçar a igualdade entre os valores. Essa diferença será considerada
na discussão das limitações e da validação dos dados.
