## Objetivo

Este notebook realiza a leitura, seleção, limpeza e preparação dos
microdados da RAIS 2025 referentes ao estado de São Paulo.

## Critérios iniciais de preparação

A partir da documentação oficial da RAIS e da inspeção preliminar
dos microdados, foram definidos os seguintes critérios iniciais:

- considerar apenas vínculos ativos em 31/12/2025;
- tratar códigos classificados como ignorados (`-1`) como valores ausentes;
- utilizar a classificação oficial de escolaridade da RAIS;
- preservar inicialmente as categorias originais de escolaridade;
- utilizar a CNAE 2.0 para classificação da atividade econômica;
- investigar valores ausentes e valores iguais a zero nas variáveis
  de remuneração antes da definição da variável salarial principal;
- não excluir observações automaticamente sem justificativa
  documentada.

```python
mapa_escolaridade = {
    1: "Analfabeto",
    2: "Até 5º ano incompleto",
    3: "5º ano completo do Fundamental",
    4: "6º ao 9º ano do Fundamental",
    5: "Fundamental completo",
    6: "Médio incompleto",
    7: "Médio completo",
    8: "Superior incompleto",
    9: "Superior completo",
    10: "Mestrado",
    11: "Doutorado"
}
```

```python
ativos_teste = ativos_teste.copy()

ativos_teste["Escolaridade"] = (
    ativos_teste["Escolaridade Após 2005 - Código"]
    .map(mapa_escolaridade)
)
```

```python
ativos_teste[
    "Escolaridade"
].value_counts(dropna=False)
```

```python
ativos_teste.groupby(
    "Escolaridade"
).agg(
    total=("Escolaridade", "size"),
    rem_dez_preenchida=("Vl Rem Dezembro Nom", "count")
)
```

```python
diagnostico_rem = (
    ativos_teste
    .groupby("Tipo Vínculo - Código")
    .agg(
        total=("Tipo Vínculo - Código", "size"),
        com_rem_dez=("Vl Rem Dezembro Nom", "count"),
        rem_media=("Vl Rem Média Nom", "mean")
    )
)

diagnostico_rem["pct_com_rem_dez"] = (
    diagnostico_rem["com_rem_dez"]
    / diagnostico_rem["total"]
    * 100
)

diagnostico_rem.sort_values(
    "total",
    ascending=False
)
```
