  entre níveis de escolaridade (H = 4.076.790,86) e entre setores

  (F = 18,78; p = 7,94e-132), mas com ganho de R² pequeno (0,5007 → 0,5020) —
  interação real, poder explicativo incremental modesto.
- **Achado adicional:** coeficiente de sexo = -21,2% (mulheres vs. homens,

## 3. Interpretação integrada

A análise estatística confirma formalmente os padrões que a EDA já sugeria: a
remuneração cresce com a escolaridade, de forma moderada até o Ensino Médio
completo e com salto expressivo a partir do Ensino Superior, e esse retorno é
maior em Indústria e Serviços do que em Comércio, Construção e Agropecuária.

A correlação moderada (rho = 0,4664) e o R² de 0,501 mostram que escolaridade e
os controles incluídos explicam cerca de metade da variação da remuneração —
uma parte relevante permanece associada a fatores não incluídos no modelo,
como a ocupação específica exercida pelo trabalhador (CBO), não incorporada por
sua alta granularidade.

A pequena não monotonicidade entre "6º ao 9º ano do Fundamental" e "5º ano
completo do Fundamental" aparece tanto na mediana bruta (EDA) quanto no
coeficiente da regressão — não é um artefato de uma das análises, mas uma
característica real dos dados nessas categorias intermediárias, com menor
número de observações.

**Resultados são associativos, não causais.** Escolaridade pode estar
correlacionada com fatores não observados (ocupação específica, seleção para
vagas melhores, capital social, entre outros).

## 4. Limitações

- **Associação, não causalidade** — desenho observacional; ocupação, idade,
  jornada, setor e tipo de vínculo também influenciam a remuneração.
- **Ocupação (CBO) não incluída como controle direto** — pode estar absorvendo
  parte do efeito atribuído à escolaridade.
- **Gap de gênero (-21,2%)** identificado mas não é o foco do trabalho — sem
  controle de ocupação, não dá para isolar quanto reflete segregação
  ocupacional vs. outros fatores.
- **Unidade de análise é o vínculo, não a pessoa** — uma mesma pessoa pode ter
  mais de um vínculo formal.
- **Regressão estimada em amostra de 300 mil vínculos** (de 14,9 milhões), por
  limitação computacional — preserva poder estatístico, mas os coeficientes são
  estimativas amostrais.
- **Escopo geográfico e temporal**: apenas SP, ano-base 2025 — resultados não
  devem ser generalizados sem qualificação para outras UFs/anos.

## 5. Conclusão e recomendação

A escolaridade está associada de forma consistente e estatisticamente robusta a
remuneração mais alta no mercado de trabalho formal de SP, com retorno que se
intensifica fortemente a partir do Ensino Superior e que é significativamente
maior em Indústria e Serviços do que em Comércio, Construção e Agropecuária.

**Recomendação:** políticas de qualificação profissional voltadas a elevar a
escolaridade — em especial viabilizar a conclusão do Ensino Superior — tendem a
estar associadas ao maior ganho relativo de remuneração observado nos dados,
sobretudo se direcionadas a trabalhadores dos setores de Indústria e Serviços.

**Ressalva:** essa recomendação é baseada em associação estatística, não em
causalidade comprovada — o modelo explica cerca de metade da variação da
remuneração, restando parcela relevante ligada a fatores não observados
(ocupação, seleção, trajetória profissional). Investigações futuras deveriam
incorporar ocupação (CBO) e aprofundar o achado sobre gênero.
